# Lesson 1 - From blink to a button-controlled LED

Each **code cell** is a normal Arduino program: globals, functions, and
optionally `setup()` / `loop()`.

**Depends on** lets a cell reuse other cells (by number or title). When you
press Upload on a cell, the app builds one sketch from:

- the globals and functions of every dependency,
- the `setup()` of every dependency, in order,
- the `loop()` of **only the cell you uploaded**.

Press **Show code** on any cell to see the combined sketch.


```c++
const int LED_PIN = 2;   // most ESP32 dev boards; change if needed

void setup() {
  pinMode(LED_PIN, OUTPUT);
}

void loop() {
  digitalWrite(LED_PIN, HIGH);
  delay(500);
  digitalWrite(LED_PIN, LOW);
  delay(500);
}
```

```c++
const int BUTTON_PIN = 4;   // button between GPIO4 and GND

void setup() {
  Serial.begin(115200);
  pinMode(BUTTON_PIN, INPUT_PULLUP);   // released = HIGH, pressed = LOW
}

void loop() {
  Serial.println(digitalRead(BUTTON_PIN));
  delay(200);
}
```

```c++
const unsigned long DEBOUNCE_MS = 30;
bool lastReading = HIGH;
bool stableState = HIGH;
unsigned long lastChange = 0;

// Returns true once per button press, ignoring contact bounce
bool buttonPressed() {
  bool reading = digitalRead(BUTTON_PIN);
  if (reading != lastReading) {
    lastChange = millis();
    lastReading = reading;
  }
  if (millis() - lastChange > DEBOUNCE_MS && reading != stableState) {
    stableState = reading;
    if (stableState == LOW) return true;
  }
  return false;
}

void loop() {
  if (buttonPressed()) Serial.println("Pressed!");
}
```

```c++
bool ledOn = false;

void setup() {
  digitalWrite(LED_PIN, LOW);
}

void loop() {
  if (buttonPressed()) {
    ledOn = !ledOn;
    digitalWrite(LED_PIN, ledOn);
    Serial.println(ledOn ? "LED ON" : "LED OFF");
  }
}
```
